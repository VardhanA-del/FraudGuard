# FraudGuard exploratory analysis
This notebook is intentionally limited to source exploration; production logic lives in `src/`.


```python
from pathlib import Path
import pandas as pd
import plotly.express as px

source_path = Path('../data/raw/creditcard.csv')
transaction_ledger = pd.read_csv(source_path)
transaction_ledger.info()
transaction_ledger['Class'].value_counts(normalize=True)
```

```python
px.histogram(transaction_ledger, x='Amount', color='Class', log_y=True, title='Amount distribution by class').show()
px.histogram(transaction_ledger, x='Time', color='Class', title='Transaction time distribution').show()
transaction_ledger.groupby('Class')['Amount'].describe()
```

## Initial observations
Fraud is rare, Amount is strongly skewed, and `Time` allows a chronological evaluation split. The anonymized `V` columns are retained without inventing customer or merchant attributes.
